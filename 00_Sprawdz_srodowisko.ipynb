## 1. Czy notebook używa właściwego Pythona?

Potrzebujemy Pythona **3.11 lub 3.12** ze środowiska `.venv` w folderze **tego repozytorium**. W VS Code otwórz folder repo i wybierz ten interpreter przez **Select Kernel**. JupyterLab uruchom z folderu repo poleceniem z instrukcji.

Poniższa komórka sprawdzi folder roboczy oraz to, czy kernel używa właściwego `.venv`. Jeśli zgłosi błąd, popraw wybór kernela lub sposób uruchomienia i wykonaj obie komórki ponownie.

```python
from pathlib import Path
import sys

print('Python:', sys.version.split()[0])
print('Interpreter:', sys.executable)

def check_notebook_kernel():
    repo_dir = Path.cwd().resolve()
    if not all((repo_dir / name).is_file() for name in
               ('check_gemini.py', '00_Sprawdz_srodowisko.ipynb')):
        raise RuntimeError('Otwórz folder tego repo w VS Code lub uruchom z niego JupyterLab. '
                           'Notebook i check_gemini.py muszą być w tym folderze.')
    if sys.version_info[:2] not in {(3, 11), (3, 12)}:
        raise RuntimeError('Wybierz kernel z Pythonem 3.11 lub 3.12.')
    expected_venv = repo_dir / '.venv'
    try:
        correct_venv = sys.prefix != sys.base_prefix and Path(sys.prefix).samefile(expected_venv)
    except OSError:
        correct_venv = False
    if not correct_venv:
        raise RuntimeError(
            f'Kernel używa: {sys.prefix}\nOczekiwane środowisko: {expected_venv}\n'
            'Utwórz .venv zgodnie z instrukcją, jeśli jeszcze go nie ma. '
            'W VS Code wybierz je przez Select Kernel. W JupyterLab: zatrzymaj serwer '
            '(Ctrl+C w terminalu) i uruchom go ponownie Pythonem z tego .venv według instrukcji.'
        )

check_notebook_kernel()
print('Kernel: PASSED')
```
